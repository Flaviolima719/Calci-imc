```python
%%html
<!DOCTYPE html>
<html lang="pt-BR">
<head>
<meta charset="UTF-8">
<meta name="viewport" content="width=device-width, initial-scale=1, viewport-fit=cover">
<title>Controle de Horas</title>
<!-- Importando Chart.js para o gráfico de barras -->
<script src="https://cdn.jsdelivr.net/npm/chart.js"></script>
<style>
:root{
  --bg: #f4f5f9;
  --card: #ffffff;
  --tx: #2d3748;
  --mu: #718096;
  --ac: #5c67f2;
  --ac-hover: #4a54e1;
  --ok: #10b981;
  --bad: #ef4444;
  --bd: #e2e8f0;
  --today-bg: rgba(92, 103, 242, 0.08);
  --today-border: #5c67f2;
  box-sizing: border-box;
  padding-top: env(safe-area-inset-top, 0px);
  padding-bottom: env(safe-area-inset-bottom, 0px);
}
@media (prefers-color-scheme: dark){
  :root:not([data-theme="light"]){
    --bg: #0f172a;
    --card: #1e293b;
    --tx: #f1f5f9;
    --mu: #94a3b8;
    --bd: #334155;
    --ac: #818cf8;
    --ac-hover: #6366f1;
    --ok: #34d399;
    --bad: #f87171;
    --today-bg: rgba(129, 140, 248, 0.12);
    --today-border: #818cf8;
  }
}
* { box-sizing: border-box; transition: background-color 0.2s, border-color 0.2s, transform 0.1s; }
body {
  margin: 0;
  background: var(--bg);
  color: var(--tx);
  font-family: system-ui, -apple-system, Segoe UI, Roboto, Helvetica, Arial, sans-serif;
  padding: 20px 16px;
  line-height: 1.5;
}
main {
  max-width: 800px;
  margin: 0 auto;
}
h1 {
  font-size: 1.8rem;
  font-weight: 800;
  margin: 0 0 16px;
  letter-spacing: -0.025em;
  background: linear-gradient(135deg, var(--ac), #ec4899);
  -webkit-background-clip: text;
  -webkit-text-fill-color: transparent;
}
.card {
  background: var(--card);
  border: 1px solid var(--bd);
  border-radius: 16px;
  padding: 18px;
  margin-bottom: 16px;
  box-shadow: 0 4px 6px -1px rgba(0, 0, 0, 0.05), 0 2px 4px -1px rgba(0, 0, 0, 0.006);
}
.card:hover {
  box-shadow: 0 10px 15px -3px rgba(0, 0, 0, 0.08), 0 4px 6px -2px rgba(0, 0, 0, 0.01);
}
.row {
  display: flex;
  gap: 12px;
  flex-wrap: wrap;
  align-items: center;
}
input, select, button {
  font: inherit;
  padding: 10px 14px;
  border-radius: 10px;
  border: 1px solid var(--bd);
  background: var(--bg);
  color: var(--tx);
  outline: none;
}
input[type=time]:focus, select:focus {
  border-color: var(--ac);
  box-shadow: 0 0 0 3px rgba(92, 103, 242, 0.25);
}
button {
  background: var(--ac);
  color: #fff;
  border: 0;
  font-weight: 600;
  cursor: pointer;
}
button:hover {
  background: var(--ac-hover);
  transform: translateY(-1px);
}
button:active {
  transform: translateY(0);
}
.btn-secondary {
  background: var(--bg);
  color: var(--tx);
  border: 1px solid var(--bd);
}
.btn-secondary:hover {
  background: var(--bd);
}
.months {
  display: flex;
  gap: 8px;
  overflow-x: auto;
  padding-bottom: 8px;
  scrollbar-width: thin;
}
.months::-webkit-scrollbar {
  height: 6px;
}
.months::-webkit-scrollbar-thumb {
  background: var(--bd);
  border-radius: 10px;
}
.months button {
  background: var(--bg);
  color: var(--tx);
  border: 1px solid var(--bd);
  flex: 0 0 auto;
  padding: 8px 14px;
}
.months button.on {
  background: var(--ac);
  color: #fff;
  border-color: var(--ac);
  box-shadow: 0 4px 12px rgba(92, 103, 242, 0.2);
}
.sum {
  display: grid;
  grid-template-columns: repeat(auto-fit, minmax(140px, 1fr));
  gap: 12px;
  text-align: center;
}
.sum > div {
  padding: 12px 8px;
  background: var(--bg);
  border-radius: 12px;
  border: 1px solid var(--bd);
}
.sum b {
  display: block;
  font-size: 1.4rem;
  font-weight: 700;
}
.sum span, .mu {
  color: var(--mu);
  font-size: 0.85rem;
  font-weight: 500;
}
.pos { color: var(--ok); }
.neg { color: var(--bad); }
.day {
  display: grid;
  grid-template-columns: 90px 1fr 1fr 90px 130px;
  gap: 8px;
  align-items: center;
  border-bottom: 1px solid var(--bd);
  padding: 10px 0;
  font-size: 0.95rem;
}
.day:last-child {
  border-bottom: none;
}
.day.off {
  opacity: 0.65;
}
.day.today {
  background: var(--today-bg);
  border-left: 4px solid var(--today-border);
  padding-left: 8px;
  border-radius: 8px;
}
.day input[type=time] {
  width: 100%;
  min-width: 0;
  padding: 6px 10px;
}
.day select.type-select {
  padding: 4px 8px;
  font-size: 0.8rem;
  border-radius: 6px;
  background: var(--card);
  cursor: pointer;
}
.free {
  grid-column: 1 / -1;
  font-size: 0.8rem;
  color: var(--mu);
  display: flex;
  align-items: center;
  gap: 6px;
  margin-top: 4px;
}
.export-actions {
  display: flex;
  gap: 8px;
  margin-top: 14px;
  justify-content: flex-end;
}
.print-header {
  display: none;
  border-bottom: 2px solid #334155;
  padding-bottom: 10px;
  margin-bottom: 20px;
}
.print-header h2 {
  margin: 0 0 4px 0;
  font-size: 1.4rem;
  color: #1e293b;
}
.print-header p {
  margin: 0;
  color: #64748b;
  font-size: 0.9rem;
}
@media(max-width: 560px){
  .day {
    grid-template-columns: 80px 1fr 1fr;
    padding: 12px 0;
  }
  .tot { grid-column: 1 / -1; }
}
h2 {
  font-size: 1.15rem;
  font-weight: 700;
  margin: 0 0 12px;
}
.chart-container {
  position: relative;
  height: 250px;
  width: 100%;
}

/* Estilos de impressão otimizados para gerar PDF limpo */
@media print {
  body {
    background: #fff !important;
    color: #000 !important;
    padding: 0 !important;
  }
  main {
    max-width: 100% !important;
  }
  .print-header {
    display: block !important;
  }
  .card {
    border: none !important;
    box-shadow: none !important;
    padding: 0 !important;
    margin-bottom: 24px !important;
    page-break-inside: avoid;
  }
  /* Esconder controles, botões e o gráfico durante a impressão */
  button, select, .months, input[type=time], .free, p.mu, .row, .chart-container, .export-actions, h1 {
    display: none !important;
  }
  /* Ajuste de colunas na versão impressa */
  .day {
    grid-template-columns: 120px 1fr 1fr 100px !important;
    border-bottom: 1px solid #ccc !important;
  }
}
</style>
</head>
<body>
<main>
<div class="print-header">
  <h2>Relatório de Controle de Horas</h2>
  <p><strong>Empregado:</strong> Flávio de Lima Santos</p>
</div>

<h1>Controle de Horas</h1>

<div class="card row">
  <label style="font-weight: 600;">Ano:
    <select id="ano" onchange="setAno(this.value)"></select>
  </label>
  <button onclick="novoAno()">+ Novo ano</button>
  <span class="mu">🎯 Meta: 6h por dia, de segunda a sexta</span>
</div>

<div class="card"><div class="months" id="meses"></div></div>

<div class="card">
  <div style="display:flex; justify-content:space-between; align-items:center; flex-wrap:wrap;">
    <h2 id="tituloMes" style="margin:0;"></h2>
    <div class="export-actions">
      <button class="btn-secondary" onclick="copiarResumo()" title="Copiar resumo em formato texto">📋 Copiar Texto</button>
      <button class="btn-secondary" onclick="window.print()" title="Exportar para PDF ou Imprimir">🖨️ Gerar PDF</button>
    </div>
  </div>
  <div class="sum" id="resumoMes" style="margin-top:14px;"></div>
</div>

<div class="card">
  <h2>📈 Saldo Mensal do Ano</h2>
  <div class="chart-container">
    <canvas id="chartSaldo"></canvas>
  </div>
</div>

<div class="card" id="dias"></div>

<div class="card">
  <h2>📊 Balanço do ano</h2>
  <div class="sum" id="resumoAno"></div>
</div>

<p class="mu" style="text-align: center; margin-top: 20px;">🔒 Os dados ficam salvos de forma privada no seu navegador. Dias úteis sem registro até hoje contam como horas devidas.</p>
</main>

<script>
var KEY='controle-horas-v2', META=360, START_DATE='2026-09-29';
var NM=['Janeiro','Fevereiro','Março','Abril','Maio','Junho','Julho','Agosto','Setembro','Outubro','Novembro','Dezembro'];
var DS=['Dom','Seg','Ter','Qua','Qui','Sex','Sáb'];
var d={anos:{}};
try{var s=localStorage.getItem(KEY);if(s)d=JSON.parse(s)}catch(e){}
function save(){try{localStorage.setItem(KEY,JSON.stringify(d))}catch(e){}}
function $(i){return document.getElementById(i)}
function p2(n){return (n<10?'0':'')+n}
var now=new Date(), hoje=now.getFullYear()+'-'+p2(now.getMonth()+1)+'-'+p2(now.getDate());
var ano=now.getFullYear(), mes=now.getMonth();
if(!d.anos[ano])d.anos[ano]={};

var saldoChart = null;

function mins(r){
  if(!r||!r.i||!r.o)return 0;
  var a=r.i.split(':'),b=r.o.split(':');
  var m=(+b[0]*60+ +b[1])-(+a[0]*60+ +a[1]);
  return m>0?m:0;
}
function fm(m){
  var s=m<0?'-':'',v=Math.abs(Math.round(m));
  return s+Math.floor(v/60)+'h'+(v%60?' '+p2(v%60)+'min':'');
}
function cls(m){return m>0?'pos':(m<0?'neg':'')}
function fmSaldo(m){return (m>0?'+':'')+fm(m)}

function stats(y,m){
  var Y=d.anos[y]||{},n=new Date(y,m+1,0).getDate(),r={devidoAteHoje:0,previsto:0,feito:0};
  for(var i=1;i<=n;i++){
    var dt=new Date(y,m,i),w=dt.getDay();
    if(w<1||w>5)continue;
    var k=y+'-'+p2(m+1)+'-'+p2(i),rec=Y[k]||{};
    var isFree = (rec.type === 'feriado' || rec.type === 'folga' || rec.type === 'nao-trabalhado' || rec.off);
    var exp = (isFree || k < START_DATE) ? 0 : META;
    r.previsto+=exp;r.feito+=mins(rec);
    if(k<=hoje)r.devidoAteHoje+=exp;
  }
  r.saldo=r.feito-r.devidoAteHoje;
  return r;
}
function box(v,l,c){return '<div><b class="'+(c||'')+'">'+v+'</b><span>'+l+'</span></div>'}

function setAno(a){ano=+a;if(!d.anos[ano])d.anos[ano]={};render()}
function novoAno(){
  var ks=Object.keys(d.anos).map(Number),sug=Math.max.apply(null,ks)+1;
  var a=parseInt(prompt('Qual ano deseja criar?',sug));
  if(!a||a<2000||a>2100)return;
  if(!d.anos[a])d.anos[a]={};
  ano=a;mes=0;save();render();
}
function setMes(m){mes=m;render()}
function setCampo(k,campo,v){
  var Y=d.anos[ano];if(!Y[k])Y[k]={};
  Y[k][campo]=v;
  if (campo === 'type') {
    Y[k]['off'] = (v === 'feriado' || v === 'folga' || v === 'nao-trabalhado');
  }
  save();render();
}

function updateChart(saldos){
  var ctx = $('chartSaldo').getContext('2d');
  var labels = NM.map(function(n){return n.slice(0,3)});
  var colors = saldos.map(function(s){return s >= 0 ? '#10b981' : '#ef4444'});
  var dataInHours = saldos.map(function(s){return Math.round((s/60)*10)/10});

  if (saldoChart) {
    saldoChart.data.datasets[0].data = dataInHours;
    saldoChart.data.datasets[0].backgroundColor = colors;
    saldoChart.data.datasets[0].borderColor = colors;
    saldoChart.update();
  } else {
    saldoChart = new Chart(ctx, {
      type: 'bar',
      data: {
        labels: labels,
        datasets: [{
          label: 'Saldo (horas)',
          data: dataInHours,
          backgroundColor: colors,
          borderColor: colors,
          borderWidth: 1,
          borderRadius: 6
        }]
      },
      options: {
        responsive: true,
        maintainAspectRatio: false,
        plugins: {
          legend: { display: false },
          tooltip: {
            callbacks: {
              label: function(context) {
                var val = saldos[context.dataIndex];
                return ' Saldo: ' + fmSaldo(val);
              }
            }
          }
        },
        scales: {
          y: {
            grid: {
              color: 'rgba(128, 128, 128, 0.15)'
            },
            ticks: {
              callback: function(value) {
                return value + 'h';
              }
            }
          },
          x: {
            grid: { display: false }
          }
        }
      }
    });
  }
}

function copiarResumo(){
  var st = stats(ano, mes);
  var texto = `📅 RESUMO MENSAL: ${NM[mes]} de ${ano}\n` +
              `-------------------------------------\n` +
              `Colaborador: Flávio de Lima Santos\n` +
              `✅ Horas Feitas: ${fm(st.feito)}\n` +
              `🎯 Exigidas até hoje: ${fm(st.devidoAteHoje)}\n` +
              `⚖️ Saldo do Mês: ${fmSaldo(st.saldo)}\n` +
              `🚀 Total Previsto do Mês: ${fm(st.previsto)}\n\n` +
              `Gerado por Controle de Horas.`;
  navigator.clipboard.writeText(texto).then(function() {
    alert('Resumo copied with success!');
  }).catch(function() {
    alert('Error copying resumen:\n\n' + texto);
  });
}

function render(){
  var anos=Object.keys(d.anos).sort();
  $('ano').innerHTML=anos.map(function(a){return '<option'+(+a===ano?' selected':'')+'>'+a+'</option>'}).join('');
  $('meses').innerHTML=NM.map(function(n,i){return '<button class="'+(i===mes?'on':'')+'" onclick="setMes('+i+')">'+n.slice(0,3)+'</button>'}).join('');
  $('tituloMes').textContent=NM[mes]+' de '+ano;

  var st=stats(ano,mes);
  $('resumoMes').innerHTML=
    box(fm(st.feito),'feitas no mês')+
    box(fm(st.devidoAteHoje),'exigidas até hoje')+
    box(st.saldo<0?fm(-st.saldo):'0h',st.saldo<0?'devendo':'em dia','neg'.slice(0,st.saldo<0?3:0))+
    (st.saldo>0?box(fm(st.saldo),'a mais (crédito)','pos'):'')+
    box(fm(st.previsto),'total do mês');

  var Y=d.anos[ano]||{},n=new Date(ano,mes+1,0).getDate(),h='';
  for(var i=1;i<=n;i++){
    var w=new Date(ano,mes,i).getDay();
    if(w<1||w>5)continue;
    var k=ano+'-'+p2(mes+1)+'-'+p2(i),rec=Y[k]||{},m=mins(rec);
    var currentType = rec.type || (rec.off ? 'feriado' : 'trabalho');
    var isFree = (currentType === 'feriado' || currentType === 'folga' || currentType === 'nao-trabalhado');
    var exp = (isFree || k < START_DATE) ? 0 : META;
    var passado=k<=hoje,saldo=m-exp;
    h+='<div class="day'+(isFree || k < START_DATE ?' off':'')+(k===hoje?' today':'')+'">'+
      '<b>'+DS[w]+' '+p2(i)+'/'+p2(mes+1)+'</b>'+
      '<span class="print-time-only" style="display:none;">'+(rec.i||'—')+' a '+(rec.o||'—')+'</span>'+
      '<input type="time" value="'+(rec.i||'')+'" onchange="setCampo(\''+k+'\',\'i\',this.value)" aria-label="Entrada" class="no-print">'+
      '<input type="time" value="'+(rec.o||'')+'" onchange="setCampo(\''+k+'\',\'o\',this.value)" aria-label="Saída" class="no-print">'+
      '<span class="tot"><b>'+(m?fm(m):'—')+'</b></span>'+
      '<span class="tot '+(isFree || k < START_DATE ||(!passado&&!m)?'':cls(saldo))+'">'+(isFree || k < START_DATE ? (currentType === 'nao-trabalhado' ? 'não trab.' : currentType) : (passado||m?fmSaldo(saldo):''))+'</span>'+
      '<div class="free no-print">'+
        '<select class="type-select" onchange="setCampo(\''+k+'\',\'type\',this.value)">'+
          '<option value="trabalho"'+(currentType==='trabalho'?' selected':'')+'>💼 Trabalho</option>'+
          '<option value="feriado"'+(currentType==='feriado'?' selected':'')+'>🎉 Feriado</option>'+
          '<option value="folga"'+(currentType==='folga'?' selected':'')+'>🏖️ Folga</option>'+
          '<option value="nao-trabalhado"'+(currentType==='nao-trabalhado'?' selected':'')+'>🚫 Não trab.</option>'+
        '</select>'+
      '</div>'+
    '</div>';
  }
  $('dias').innerHTML='<div class="mu no-print" style="margin-bottom:10px; font-weight:600;">Dia &middot; Entrada &middot; Saída &middot; Feitas &middot; Saldo &middot; Categoria</div>'+h;

  var sheet = document.createElement('style');
  sheet.innerHTML = "@media print { .no-print { display:none !important; } .print-time-only { display:inline !important; } }";
  document.body.appendChild(sheet);

  var f=0,dv=0,pv=0;
  var saldosAnuais = [];
  for(var m2=0;m2<12;m2++){
    var s2=stats(ano,m2);
    f+=s2.feito;
    dv+=s2.devidoAteHoje;
    pv+=s2.previsto;
    saldosAnuais.push(s2.saldo);
  }
  var sa=f-dv;
  $('resumoAno').innerHTML=
    box(fm(f),'feitas no ano')+
    box(fm(dv),'exigidas até hoje')+
    box(sa<0?fm(-sa):(sa>0?fmSaldo(sa):'0h'),sa<0?'devendo':(sa>0?'crédito':'em dia'),cls(sa))+
    box(fm(pv),'total previsto do ano');

  updateChart(saldosAnuais);
}

function setCampo(k,campo,v){
  var Y=d.anos[ano];if(!Y[k])Y[k]={};
  Y[k][campo]=v;
  if (campo === 'type') {
    Y[k]['off'] = (v === 'feriado' || v === 'folga' || v === 'nao-trabalhado');
  }
  save();render();
}

save();render();
</script>
</body>
</html>
```
